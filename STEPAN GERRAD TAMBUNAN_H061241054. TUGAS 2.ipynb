# **1. Pendahuluan**  
Artificial Intelligence (AI) merupakan teknologi yang dapat membantu mahasiswa geofisika dalam memahami konsep, mengolah data, membuat visualisasi, menulis program Python, dan membantu interpretasi data geofisika.

Salah satu kerangka yang dapat digunakan adalah **R–T–C–I–K–O**, yaitu Role, Task, Context, Input, Constraints, dan Output.


Namun, hasil yang diberikan AI sangat bergantung pada instruksi atau prompt yang digunakan. Prompt yang kurang jelas dapat menghasilkan jawaban yang terlalu umum, perhitungan yang tidak sesuai, atau interpretasi yang tidak didukung oleh data.

Oleh karena itu, diperlukan prompt yang terstruktur agar AI dapat memahami tugas, data, batasan analisis, dan bentuk jawaban yang diharapkan.

AI sebaiknya digunakan sebagai alat bantu, bukan sebagai pengganti pemeriksaan ilmiah. Setiap hasil analisis tetap harus diperiksa berdasarkan data, persamaan, satuan, dan teori geofisika yang relevan.

# **2. Framework R–T–C–I–K–O**

| Komponen | Kepanjangan | Fungsi                                   |
| -------- | ----------- | ---------------------------------------- |
| R        | Role        | Menentukan peran AI.                     |
| T        | Task        | Menentukan tugas yang harus dikerjakan.  |
| C        | Context     | Memberikan latar belakang dan teori.     |
| I        | Input       | Menjelaskan data yang tersedia.          |
| K        | Constraints | Menentukan batasan analisis.             |
| O        | Output      | Menentukan bentuk hasil yang diharapkan. |

## **2.1 Role**

Role merupakan bagian prompt yang menjelaskan peran atau kompetensi yang diharapkan dari AI.

Contoh:

> Anda berperan sebagai pendamping analisis geofisika yang memahami metode geolistrik, seismik, GPR, dan geomagnet.

Dengan menentukan role, jawaban AI diharapkan lebih sesuai dengan bidang yang sedang dipelajari.

## **2.2 Task**

Task menjelaskan pekerjaan yang harus dilakukan oleh AI.

Contoh:

> Analisis data geolistrik, identifikasi perubahan resistivitas, dan jelaskan kemungkinan karakter material bawah permukaan.

Task sebaiknya dibuat secara spesifik agar AI memahami tujuan analisis.
## **2.3 Context**

Context merupakan informasi latar belakang yang membantu AI memahami permasalahan.

Contoh:

> Metode geolistrik memanfaatkan perbedaan sifat kelistrikan material bawah permukaan untuk membantu mengidentifikasi variasi lapisan atau zona tertentu.

## **2.4 Input**

Input berisi data yang akan dianalisis.

Contohnya berupa:

* Nilai resistivitas.
* Waktu tempuh gelombang seismik.
* Kecepatan gelombang.
* Data radargram.
* Nilai anomali magnetik.
* Nilai chargeability.
* Informasi geologi pendukung.

Data yang digunakan harus jelas dan memiliki satuan apabila diperlukan.

## **2.5 Constraints**

Constraints merupakan batasan yang harus dipatuhi oleh AI.

Contohnya:

* Tidak mengarang data.
* Tidak mengubah nilai pengukuran.
* Memeriksa satuan sebelum melakukan perhitungan.
* Menggunakan persamaan yang sesuai.
* Memisahkan hasil pengamatan dan interpretasi.
* Menyebutkan keterbatasan data.

## **2.6 Output**

Output menjelaskan bentuk jawaban yang diharapkan.

Contohnya:
* Penjelasan dalam bentuk paragraf.
* Tabel hasil analisis.
* Grafik atau penampang.
* Perhitungan langkah demi langkah.

# **4. Contoh Penerapan Prompt**

Misalnya, diketahui:
* Kecepatan gelombang: 2.000 m/s.
* Waktu tempuh dua arah atau Two-Way Travel Time (TWT): 500 ms.

Langkah pertama adalah mengubah waktu dari milidetik menjadi detik.

$$
TWT = 500 \text{ ms} = 0,5 \text{ s}
$$

Kedalaman reflektor dapat dihitung menggunakan persamaan sederhana:

$$
Z = \frac{V \times TWT}{2}
$$

Keterangan:

* \(Z\) = kedalaman reflektor (m).
* \(V\) = kecepatan gelombang (m/s).
* \(TWT\) = waktu tempuh dua arah (s).

Substitusi nilai:

$$
Z = \frac{2000 \times 0,5}{2}
$$

$$
Z = 500 \text{ m}
$$

Dengan asumsi kecepatan tersebut mewakili lintasan gelombang, kedalaman reflektor diperkirakan sekitar 500 meter.

Dalam kondisi nyata, ketidakpastian kecepatan dan variasi lapisan bawah permukaan dapat memengaruhi hasil perhitungan.

## **6.1 Studi Kasus Geolistrik**

### **Deskripsi**

Metode geolistrik digunakan untuk mempelajari variasi resistivitas bawah permukaan. Perbedaan resistivitas dapat berkaitan dengan jenis batuan, porositas, kandungan air, salinitas, dan kondisi geologi.

### **Prompt**

**Role:** Anda berperan sebagai pendamping analisis data geolistrik.

**Task:** Analisis pola resistivitas dan jelaskan variasi karakter bawah permukaan.

**Context:** Resistivitas merupakan sifat kelistrikan material yang dipengaruhi oleh komposisi, porositas, fluida, dan kondisi geologi.

**Input:** Gunakan data resistivitas, penampang, kedalaman, dan informasi geologi jika tersedia.

**Constraints:** Jangan menentukan jenis batuan atau keberadaan air tanah hanya berdasarkan nilai resistivitas tanpa mempertimbangkan data pendukung.

**Output:** Jelaskan pola resistivitas rendah dan tinggi, distribusi kedalaman, serta kemungkinan interpretasi dan keterbatasannya.

### **Interpretasi Contoh**

Pada model sintetis, zona resistivitas rendah berada di bagian tengah penampang, sedangkan zona resistivitas tinggi berada di bagian kanan pada kedalaman tertentu.

Resistivitas rendah dapat berkaitan dengan material lempung atau material yang mengandung fluida lebih konduktif. Resistivitas tinggi dapat berkaitan dengan material yang lebih resisif.
Namun, hubungan tersebut tidak bersifat mutlak. Interpretasi yang lebih pasti membutuhkan informasi geologi dan data pendukung.


````python
import numpy as np
import matplotlib.pyplot as plt

# Mengatur ukuran grafik
plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["axes.grid"] = True

print("Library berhasil diimpor.")
````

````python
import numpy as np
import matplotlib.pyplot as plt

# Membuat koordinat lintasan dan kedalaman
x = np.linspace(0, 120, 121)
z = np.linspace(0, 30, 61)
X, Z = np.meshgrid(x, z)

# Membuat model resistivitas awal
rho = 110 + 0.7 * Z + 0 * X

# Membuat zona resistivitas rendah
anomali = (X > 42) & (X < 78) & (Z > 7) & (Z < 22)
rho[anomali] = 30

# Membuat visualisasi
fig, ax = plt.subplots(figsize=(9, 4.5))

im = ax.pcolormesh(
    X, Z, rho,
    shading="auto",
    cmap="viridis"
)

ax.invert_yaxis()
ax.set_title("Penampang Resistivitas Sintetis")
ax.set_xlabel("Jarak (m)")
ax.set_ylabel("Kedalaman (m)")

fig.colorbar(
    im,
    ax=ax,
    label="Resistivitas (ohm.m)"
)

plt.tight_layout()
plt.show()
````

## 6.2 GPR — Ilustrasi Atenuasi dan Pilihan Frekuensi

Kode ini membandingkan kurva amplitudo terhadap kedalaman untuk beberapa frekuensi antena menggunakan model atenuasi sederhana. Model hanya menunjukkan konsep umum bahwa sinyal dapat melemah lebih cepat pada frekuensi tinggi; penetrasi nyata sangat dipengaruhi konduktivitas, permitivitas, kadar air, antena, dan kondisi akuisisi.

````python

import numpy as np
import matplotlib.pyplot as plt

depth = np.linspace(0, 10, 300)  # kedalaman ilustratif (m)
frequencies = [100, 250, 500]   # MHz
# Konstanta atenuasi ilustratif per meter; bukan hasil pengukuran
attenuation = {100: 0.22, 250: 0.42, 500: 0.72}

plt.figure(figsize=(9, 5))
for f in frequencies:
    amp = np.exp(-attenuation[f] * depth)
    plt.plot(depth, amp, linewidth=2, label=f"{f} MHz")
    depth_10pct = np.log(10) / attenuation[f]
    print(f"{f} MHz: kedalaman saat amplitudo turun menjadi 10% ≈ {depth_10pct:.2f} m (model ilustratif)")
plt.xlabel("Kedalaman ilustratif (m)")
plt.ylabel("Amplitudo relatif")
plt.title("Ilustrasi Atenuasi GPR untuk Beberapa Frekuensi")
plt.ylim(0, 1.02)
plt.grid(True, alpha=0.3)
plt.legend(title="Frekuensi antena")
plt.tight_layout()
plt.show()

# Contoh radargram sintetis sederhana dengan atenuasi terhadap waktu
rng = np.random.default_rng(23)
x = np.linspace(0, 20, 161)
t = np.linspace(0, 100, 201)  # ns
X, T = np.meshgrid(x, t)
radargram = 0.10 * rng.normal(size=X.shape)
radargram += np.exp(-((T-15)/2.5)**2) * np.cos(2*np.pi*(T-15)/7)
radargram += 0.65*np.exp(-((T-38)/3.5)**2) * np.cos(2*np.pi*(T-38)/8)
radargram *= np.exp(-T/55)  # pelemahan amplitudo terhadap waktu

plt.figure(figsize=(9, 4.5))
plt.pcolormesh(X, T, radargram, shading="auto", cmap="gray")
plt.gca().invert_yaxis()
plt.xlabel("Jarak lintasan (m)")
plt.ylabel("Two-way travel time (ns)")
plt.title("Contoh Radargram Sintetis dengan Atenuasi")
plt.colorbar(label="Amplitudo relatif")
plt.tight_layout()
plt.show()
````

## 6.3 Seismik Refraksi — Analisis Travel Time dan Ambiguitas Lapisan

Kode ini membuat data waktu tiba pertama (*first arrival*) sintetis, menambahkan sedikit noise, menampilkan kurva waktu tempuh terhadap jarak, lalu memperkirakan kecepatan semu dari kemiringan (*slope*) pada beberapa segmen. Perubahan slope dapat mengindikasikan perubahan kecepatan, tetapi tidak secara otomatis membuktikan batas lapisan.

````python

import numpy as np
import matplotlib.pyplot as plt

# Data sintetis: jarak geophone (m) dan first-arrival time (ms)
rng = np.random.default_rng(21)
x = np.arange(0, 101, 5, dtype=float)

# Dua segmen waktu tempuh ideal (ilustrasi dua kemiringan)
t_ideal = np.where(x <= 40, 2.0 + (x / 500.0) * 1000,
                   82.0 + ((x - 40) / 1200.0) * 1000)
t_obs = t_ideal + rng.normal(0, 1.2, size=x.size)

# Estimasi slope secara terpisah untuk dua segmen
segments = [(x <= 40, "Segmen 1"), (x > 40, "Segmen 2")]
print("SEISMIK REFRAKSI — DATA SINTETIS")
print(f"Jumlah geophone: {len(x)}")
print("Catatan: estimasi kecepatan sederhana dari slope; bukan inversi refraksi lengkap.")

fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(x, t_obs, label="First arrival sintetis", color="black", s=28)
colors = ["tab:blue", "tab:orange"]
for (mask, label), color in zip(segments, colors):
    if mask.sum() >= 2:
        slope, intercept = np.polyfit(x[mask], t_obs[mask], 1)
        x_fit = x[mask]
        ax.plot(x_fit, slope*x_fit + intercept, color=color, linewidth=2, label=label)
        # t dalam ms, x dalam m: V (m/s) = 1000 / slope(ms/m)
        velocity = 1000.0 / slope if slope > 0 else np.nan
        print(f"{label}: slope = {slope:.3f} ms/m; kecepatan semu ≈ {velocity:.1f} m/s")
ax.set_xlabel("Jarak geophone (m)")
ax.set_ylabel("First-arrival time (ms)")
ax.set_title("Kurva Travel Time–Distance Seismik Refraksi")
ax.grid(True, alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()
````

## **6.4 Studi Kasus Geomagnet**

### **Deskripsi**

Metode geomagnet mengukur variasi medan magnet bumi yang dapat dipengaruhi oleh perbedaan sifat kemagnetan material bawah permukaan.

Data biasanya diolah menjadi peta anomali magnetik untuk membantu mengidentifikasi variasi sumber magnetik.

### **Prompt**

**Role:** Anda berperan sebagai pendamping analisis data geomagnet.

**Task:** Analisis distribusi anomali magnetik dan jelaskan pola anomali yang terlihat.

**Context:** Anomali magnetik dapat berkaitan dengan variasi kemagnetan batuan dan geometri sumber di bawah permukaan.

**Input:** Gunakan data anomali magnetik, koordinat, satuan, dan informasi geologi pendukung.

**Constraints:** Jangan langsung menyimpulkan jenis batuan atau keberadaan mineral tertentu hanya berdasarkan tinggi-rendah anomali.

**Output:** Jelaskan pola anomali, lokasi nilai relatif tinggi dan rendah, serta kemungkinan interpretasi.

### **Interpretasi Contoh**

Peta kontur sintetis menunjukkan variasi anomali magnetik secara horizontal. Zona dengan nilai relatif tinggi dan rendah dapat menunjukkan adanya perbedaan respons magnetik dalam model.

Interpretasi sumber anomali yang sebenarnya memerlukan pengolahan lebih lanjut dan informasi geologi pendukung.

````python
import numpy as np
import matplotlib.pyplot as plt

# Membuat koordinat survei
x = np.linspace(0, 100, 101)
y = np.linspace(0, 100, 101)

X, Y = np.meshgrid(x, y)

# Membuat anomali magnetik positif
anomali_positif = 100 * np.exp(
    -((X - 35) ** 2 + (Y - 60) ** 2) / 200
)

# Membuat anomali magnetik negatif
anomali_negatif = -70 * np.exp(
    -((X - 70) ** 2 + (Y - 35) ** 2) / 300
)

# Menggabungkan anomali
magnetik = anomali_positif + anomali_negatif

# Membuat peta kontur
plt.figure(figsize=(9, 7))

kontur = plt.contourf(
    X, Y, magnetik,
    levels=20,
    cmap="RdBu_r"
)

plt.colorbar(kontur, label="Anomali magnetik relatif")

plt.contour(
    X, Y, magnetik,
    levels=10,
    colors="black",
    linewidths=0.5
)

plt.xlabel("Koordinat X (m)")
plt.ylabel("Koordinat Y (m)")
plt.title("Peta Kontur Anomali Magnetik Sintetis")

plt.tight_layout()
plt.show()

print("Interpretasi:")
print("Model menunjukkan zona anomali magnetik relatif positif dan negatif.")
print("Pola anomali menggambarkan variasi respons magnetik dalam model.")
print("Identifikasi sumber anomali memerlukan analisis dan data pendukung.")
````

## **6.5 Studi Kasus Induced Polarization (IP)**

### **Deskripsi**

Metode Induced Polarization atau IP digunakan untuk mempelajari respons polarisasi material bawah permukaan. Salah satu parameter yang dianalisis adalah chargeability, yaitu ukuran respons polarisasi material setelah arus listrik diberikan.

Metode IP sering digunakan bersama geolistrik untuk membantu karakterisasi bawah permukaan, termasuk pada survei mineralisasi.

### **Prompt**

**Role:** Anda berperan sebagai pendamping analisis data Induced Polarization.

**Task:** Analisis distribusi resistivitas dan chargeability.

**Context:** Resistivitas menunjukkan respons kelistrikan material, sedangkan chargeability menggambarkan respons polarisasinya.

**Input:** Gunakan penampang resistivitas, chargeability, kedalaman, dan informasi geologi jika tersedia.

**Constraints:** Jangan memastikan adanya mineralisasi hanya berdasarkan anomali chargeability. Pertimbangkan kemungkinan respons dari material lain.

**Output:** Jelaskan pola kedua parameter, hubungan spasialnya, dan kemungkinan interpretasi dengan keterbatasan yang jelas.

### **Interpretasi Contoh**

Pada model sintetis, zona chargeability tinggi dapat dibuat berimpit dengan zona resistivitas tertentu. Pola tersebut digunakan untuk menunjukkan bagaimana dua parameter dapat dianalisis secara bersamaan.

Dalam survei nyata, anomali chargeability perlu dievaluasi bersama resistivitas, geologi, data geokimia, dan informasi pendukung lainnya.

````python
import numpy as np
import matplotlib.pyplot as plt

# Membuat koordinat
x = np.linspace(0, 100, 101)
z = np.linspace(0, 30, 61)

X, Z = np.meshgrid(x, z)

# Model resistivitas latar belakang
resistivitas = np.full_like(X, 100.0)

# Zona resistivitas rendah
resistivitas[
    (X >= 35) & (X <= 60) &
    (Z >= 8) & (Z <= 22)
] = 25.0

# Zona resistivitas tinggi
resistivitas[
    (X >= 70) & (Z >= 10)
] = 250.0

# Model chargeability latar belakang
chargeability = np.full_like(X, 5.0)

# Membuat zona chargeability tinggi
chargeability[
    (X >= 35) & (X <= 60) &
    (Z >= 8) & (Z <= 22)
] = 25.0

# Membuat dua grafik
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# Grafik resistivitas
gambar1 = ax[0].pcolormesh(
    X, Z, resistivitas,
    shading="auto",
    cmap="viridis"
)

ax[0].set_title("Penampang Resistivitas")
ax[0].set_xlabel("Jarak (m)")
ax[0].set_ylabel("Kedalaman (m)")
ax[0].invert_yaxis()

fig.colorbar(
    gambar1,
    ax=ax[0],
    label="Resistivitas (Ohm.m)"
)

# Grafik chargeability
gambar2 = ax[1].pcolormesh(
    X, Z, chargeability,
    shading="auto",
    cmap="plasma"
)

ax[1].set_title("Penampang Chargeability")
ax[1].set_xlabel("Jarak (m)")
ax[1].set_ylabel("Kedalaman (m)")
ax[1].invert_yaxis()

fig.colorbar(
    gambar2,
    ax=ax[1],
    label="Chargeability relatif"
)

plt.suptitle("Model Sintetis Metode Induced Polarization")
plt.tight_layout()
plt.show()

print("Interpretasi:")
print("Zona chargeability tinggi dibuat pada bagian tengah model.")
print("Resistivitas dan chargeability perlu dianalisis secara bersamaan.")
print("Chargeability tinggi tidak selalu membuktikan adanya mineralisasi.")
````
